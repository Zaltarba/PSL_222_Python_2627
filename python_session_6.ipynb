## Pull Stock Data using yfinance

Let’s connect to real-time stock data.

```python
import yfinance as yf

stock = st.text_input("Enter stock ticker (e.g. AAPL):")

if st.button("Load data"):
    data = yf.Ticker(stock)
    hist = data.history(period="1mo")
    st.line_chart(hist['Close'])
```

Change the time period from "1mo" to "3mo".
Add a date picker or selectbox to let the user choose the time range.
